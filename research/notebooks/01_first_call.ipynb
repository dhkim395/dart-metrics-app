```python
import os
import requests
from dotenv import load_dotenv

# .env 파일 로드 (notebooks의 상위 폴더에 있음)
load_dotenv(dotenv_path="../.env")

api_key = os.getenv("DART_API_KEY")
print(f"API 키 로드: {'성공 ✅' if api_key else '실패 ❌'}")
print(f"키 길이: {len(api_key) if api_key else 0}자 (40자여야 정상)")
```

```python
# 삼성전자 기업 정보 조회
# 삼성전자 corp_code: 00126380 (DART 고유번호)
url = "https://opendart.fss.or.kr/api/company.json"
params = {
    "crtfc_key": api_key,
    "corp_code": "00126380"
}

response = requests.get(url, params=params)
data = response.json()

print(f"응답 상태: {data.get('status')}")  # '000' = 성공
print(f"메시지: {data.get('message')}")
print()
print(f"회사명: {data.get('corp_name')}")
print(f"종목코드: {data.get('stock_code')}")
print(f"대표자: {data.get('ceo_nm')}")
print(f"업종: {data.get('induty_code')}")
print(f"설립일: {data.get('est_dt')}")
print(f"홈페이지: {data.get('hm_url')}")
```
